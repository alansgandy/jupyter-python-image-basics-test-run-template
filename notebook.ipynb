## Objectives
- Load and display an image using OpenCV.
- Convert the image to grayscale.
- Resize the image.

```python
# Import necessary libraries
import cv2
import numpy as np
```

## Step 1: Load and Display an Image

In this step, you will load an image from a file and display it using OpenCV.

### TODO
1. Use `cv2.imread()` to load the image `sample.jpg`.
2. Use `cv2.imshow()` to display the image in a window.
3. Use `cv2.waitKey(0)` and `cv2.destroyAllWindows()` to handle window events.


```python
# TODO: Load and display the image
# 1. Load the image using cv2.imread()
image = cv2.imread('sample.jpg')
# 2. Display the image using cv2.imshow()
cv2.imshow('Image', image)
# 3. Use cv2.waitKey(0) and cv2.destroyAllWindows() to handle window events
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Step 2: Convert the Image to Grayscale

Convert the loaded image to grayscale using OpenCV.

### TODO
1. Use `cv2.cvtColor()` with the appropriate flag to convert the image to grayscale.
2. Display the grayscale image using `cv2.imshow()`.
3. Handle window events with `cv2.waitKey(0)` and `cv2.destroyAllWindows()`.


```python
# TODO: Convert the image to grayscale
# 1. Convert the image to grayscale using cv2.cvtColor()
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
# 2. Display the grayscale image using cv2.imshow()
cv2.imshow('Grayscale Image', gray_image)
# 3. Use cv2.waitKey(0) and cv2.destroyAllWindows() to handle window events
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Step 3: Resize the Image

Resize the grayscale image to a new size.

### TODO
1. Use `cv2.resize()` to resize the grayscale image to 100x100 pixels.
2. Display the resized image using `cv2.imshow()`.
3. Handle window events with `cv2.waitKey(0)` and `cv2.destroyAllWindows()`.


```python
# TODO: Resize the image
# 1. Resize the grayscale image to 100x100 pixels using cv2.resize()
resized_image = cv2.resize(gray_image, (100, 100))
# 2. Display the resized image using cv2.imshow()
cv2.imshow('Resized Image', resized_image)
# 3. Use cv2.waitKey(0) and cv2.destroyAllWindows() to handle window events
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Step 4: Save the Processed Image

Save the resized grayscale image to a file.

### TODO
1. Use `cv2.imwrite()` to save the resized image as `processed_image.jpg`.


```python
# TODO: Save the processed image
# 1. Save the resized image using cv2.imwrite()
cv2.imwrite('processed_image.jpg', resized_image)
```

## Final Step: Verify Your Results

Run the following cell to verify that your image processing steps were successful. This cell should display the final processed image.


```python
# Verify the results
# Load the processed image
processed_image = cv2.imread('processed_image.jpg')
# Display the processed image
cv2.imshow('Processed Image', processed_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
